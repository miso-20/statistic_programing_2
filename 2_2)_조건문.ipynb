# 조건문
- 조건문은 **조건의 참/거짓에 따라 실행할 코드를 선택**할 때 사용하는 중요한 제어문이다.
- Python에서는 주로 `if, else`, `elif`를 사용한다.
- ** `if` 뒤에는 조건식이 오며, Python은 그 값을 참(`True`) 또는 거짓(`False`)으로 판단한다. **
- `else`와 `elif`는 필요할 때만 사용하며, 항상 작성해야 하는 것은 아니다.

## 1. `if` 문
기본 형태는 다음과 같다.

    if 조건식:
        실행문 1
    else:
        실행문 2

- 조건식이 참이면 `실행문 1`이 실행되고, 거짓이면 `실행문 2`가 실행된다.
- 다음은 가장 간단한 예시이다.

```python
a = True

if a:
    print("True 값이다.")
else:
    print("False 값이다.")
```

```python
a = False

if a:
    print("True 값이다.")
else:
    print("False 값이다.")
```

- `if`, `elif`, `else`에 속하는 문장은 반드시 **들여쓰기**해야 한다. **
- Python에서는 일반적으로 **공백 4칸**을 권장한다. 탭과 공백을 섞어 사용하면 오류가 발생할 수 있으므로 한 가지 방식으로 통일한다.


## 2. 비교 연산자
- 비교 연산자는 두 값을 비교하여 `True` 또는 `False`를 만든다.
- 값의 크기나 동등 여부를 비교하는 연산자는 `==`, `!=`, `>`, `>=`, `<`, `<=`이다.
    1) `x == y`: x와 y가 같냐?
    2) `x != y`: x와 y가 같지 않냐?
    3) `x > y`: x가 y보다 크냐?
    4) `x >=y `: x가 y보다 크거나 같냐?
    5) `x < y`: x가 y보다 작냐?
    6) `x <= y`: x가 y보다 작거나 같냐/
- 특히 `=`는 **대입 연산자**, `==`는 **같은지 비교하는 연산자**이므로 구분해야 한다.
- 비교할 때는 **값뿐 아니라 자료형도 중요**하다. 예를 들어 문자열 `"0"`과 정수 `0`은 사람이 보기에는 비슷하지만 서로 다른 자료형이므로 `==`로 비교하면 `False`가 된다.

```python
"0" == 0
```

- 비교 연산자와 나머지 연산자 `%`를 이용하면 정수가 짝수인지 홀수인지 판별할 수 있다.


```python
a = 10

if a % 2 == 0:
    print("짝수다")
else:
    print("홀수다")
```

```python
a = 9

if a % 2 == 0:
    print("짝수다")
else:
    print("홀수다")
```

- 비교 연산자는 숫자뿐만 아니라 문자열에도 사용할 수 있다.
- 문자열의 대소 비교는 왼쪽 문자부터 순서대로 비교한다.
- 영문은 일반적으로 알파벳 순서로 비교하며, 대문자와 소문자를 비교할 경우 대문자가 앞선다.

## 3. 논리 연산자: `and`, `or`, `not`
- 여러 조건을 조합할 때 `and`, `or`, `not`을 사용한다.
    1) `A and B`: A와 B가 모두 참일 때 참
    2) `A or B`: A와 B 중 하나 이상이 참일 때 참
    3) `not A`: A의 참/거짓을 반대로 바꿈
- 예를 들어 어떤 수가 **2의 배수이면서 3의 배수**, 즉 2와 3의 공배수인지 판단하려면 `and`를 사용한다.

```python
# 1. and   **** 통째로 암기 ****
# A and B ->
# T and T -> T
# T and F -> F
# F and T -> F
# F and F -> F

# 2. or
# T or T -> T
# T or F -> T
# F or T -> T
# F or F -> F
```

```python
a = 6

if (a % 2 == 0) and (a % 3 == 0):
    print("2와 3의 공배수")
else:
    print("2와 3의 공배수가 아님")
```

- 반대로 **2의 배수 또는 3의 배수**인지 판단하려면 `or`를 사용한다.


```python
a = 9

if (a % 2 == 0) or (a % 3 == 0):
    print("2 또는 3의 배수")
else:
    print("2와 3의 배수가 모두 아님")
```

```python
a = 12

if (a % 2 == 0) or (a % 3 == 0):
    print("2 또는 3의 배수")
else:
    print("2와 3의 배수가 모두 아님")
```

- `not`은 조건의 결과를 반대로 바꿀 때 사용한다.


```python
a = 5

if not (a < 0): # not은 잘 안씀  ## 시험 X
    print("0 이상")
else:
    print("음수")
```

## 4. 포함 여부 연산자: `in`, `not in` -> 안에 있냐? 없냐?
- 문자열, 리스트, 튜플 등에서 특정 값이 포함되어 있는지 확인할 때 사용하는 연산자이다.
    - x in list, x in tuple, x in 문자열: x가 list, tuple, 문자열에 있냐?
    - x not in 리스트, x not in tuple, x not in 문자열: x가 list, tuple, 문자열에 없냐?

```python
print(1 in [1, 2, 3])
print(1 not in [1, 2, 3])
```

- 첫 번째 식은 리스트 `[1, 2, 3]` 안에 `1`이 있냐? 에 대해 존재하기에 `True`를 반환한다.
- 두 번째 식은 리스트 `[1, 2, 3]` 안에 `1`이 없냐? 에 대해 존재하기에 `False`를 반환한다.
- 튜플과 문자열에도 같은 방식으로 사용할 수 있다.

```python
print("a" in ("a", "b", "c"))
print("j" not in "python")
```

- 문자열을 비교하거나 문자열에서 포함 여부를 확인할 때는 **대소문자를 구분**한다.


```python
print("A" in ["a", "b"])
```

- `in`을 조건식에 사용한 예시는 다음과 같다. 단어에 `a`가 포함되어 있으면 `"a 포함"`, 포함되어 있지 않으면 `"a 없음"`을 출력한다.

```python
x = "point"

if "a" in x:
    print("a 포함")
else:
    print("a 없음")
```

## 5. `elif` 구문
- 조건에 따라 **여러 갈래 중 하나를 선택**해야 할 때 `elif`를 사용한다.
- 같은 동작을 중첩된 `if`로도 작성할 수 있지만, 서로 배타적인 여러 조건을 차례대로 검사할 때는 `elif`가 더 간결하다.
- 예를 들어 10 이상이면 `"10보다 크거나 같음"`, 0 미만이면 `"음수"`, 그 외에는 `"음수는 아니지만 10보다 작음"`을 출력한다고 하자. 중첩된 `if`로 작성하면 다음과 같다.

- ** x = 100개의 값, 그 중 10보다 큰 값이 10개. **

```python
x = -4

if x >= 10:    # -> 100개 비교함 # 위의 ** 예시
    print("10보다 크거나 같음")
else:
    if x < 0:  # -> 90개 비교함
        print("음수")
    else:
        print("음수는 아니지만 10보다 작음")
```

```python
x = -4

if x >= 10:    # -> 100개 비교함
    print("10보다 크거나 같음")

if x < 0:      # -> 100개 비교함   # 바로 위의 if 와 다른 if
        print("음수")
else:
        print("음수는 아니지만 10보다 작음")
```

- 같은 코드를 `elif`를 사용하여 작성하면 다음과 같다.

```python
x = -5

if x >= 10:    # -> 100개 비교함
    print("10보다 크거나 같음")
elif x < 0:    # -> 90개 비교함
    print("음수")
else:
    print("음수는 아니지만 10보다 작음")
```

    - ** `False`, `None`, 숫자 `0`, 빈 문자열 `""`, 빈 list `[]`, 빈 tuple `()`, 빈 dictionary `{}` ** -> False로 판단
- 그 외의 대부분의 값은 참으로 판단한다. 예를들어, 다음과 같다.
    - ***** `1`, `-1`, `3.14`, `Python`, `[1, 2, 3]`, `[False, False]` ***** -> True로 판단

- *** `[False]`, `[False, True]`, `[False, False]`처럼 내부에 `False`가 있더라도 비어 있지 않으면 참으로 판단한다. ***

#### 1. `and`, `or`의 실제 동작
- `and`와 `or`은 두 값을 무조건 `True` 또는 `False`로 변환해서 반환하는 연산자가 아니다.
- 피연산자의 `True/False`을 판단한 뒤, 피연산자중 하나를 그대로 반환한다.
- `A and B` 연산을 실행하면 다음과 같이 진행된다.
    - `A`의 `True/False`를 판단한다.
    - `A`가 `False`이면 `A`를 그대로 반환한다.
    - `A`가 `True`이면 `B`를 그대로 반환한다.

```python
print(1 and 25)   # 1이 True니까  B인 25 반환  # and니까 앞에꺼 true면 뒤에 것도 계산해봐야 하니까 뒤에꺼 반환 ** input / output 명확하게 이해 **
print(0 and 25)   # 0이 False니까 A인 0 반환   # and인데 뒤에 것까지 계산할 필요없으니까 앞에꺼 반환
```

- `1`은 `True`이므로, 두 번째 값 `25`를 반환한다.
- `0`은 `False`이므로, 첫 번째 값 `0`을 반환한다.
- 이는 `list`에서도 동일하게 작동한다.

```python
print([True, False, True] and [True, True])  # True and True
print([] and [True, True])                   # False and ____ => [] 그대로 반환
```

- 첫 번째 예시: 첫 번째 list `[True, False, True]`가 비어 있지 않기 때문에 `True`로 판단하고, 두 번째 list `[True, True]`를 그대로 반환한다.
- 두 번째 예시: 첫 번째 list `[]`는 비어 있기 때문에 `False`로 판단하고, 첫 번째 list `[]`를 그대로 반환한다.
- `A or B`연산은 다음과 같이 진행된다.
    - `A`의 `True/False`를 판단한다.
    - `A`가 참이면 `A`를 그대로 반환한다.
    - `A`가 거짓이면 `B`를 그대로 반환한다.

```python
print(1 or 25)  # 앞에꺼가 True니까 or에선 어차피 true임 -> 앞에꺼 반환
print(0 or 25)  # 앞에꺼가 False니까 or에선 뒤의 것도 비교해야 함 -> 뒤에꺼 반환
```

- `1`은 `True`이므로 첫 번째 값 `1`을 반환한다.
- `0`은 `False`이므로 두 번째 값 `25`를 반환한다.
- 이는 `list`에서도 동일하게 작동한다.

```python
print([1, 2] or [3, 4])  # 앞에꺼가 True니까 or에선 어차피 true임 -> 앞에꺼 반환
print([] or [3, 4])      # 앞에꺼가 False니까 or에선 뒤의 것도 비교해야 함 -> 뒤에꺼 반환
```

- 첫 번째 예시: 첫 번째 list `[1, 2]`가 비어있지 않기 때문에 `True`로 판단하며, 첫 번째 list `[1, 2]`를 반환한다.
- 두 번째 예시: 첫 번째 list `[]`가 비어있기 때문에 `False`로 판단하며, 두번째 list `[3, 4]`를 반환한다.

#### 2. Short-Circuit evaluation (단락 평가)
- `and`와 `or`은 결과가 이미 결정되면 뒤의 피연산자를 평가하지 않는다.
    - `A and B`: `A`가 `False`이면 `B`를 평가하지 않는다. `A`가 `True`이면 `B`를 평가한다.
    - `A or B`: `A`가 `True`이면 `B`를 평가하지 않는다. `A`가 `False`이면 `B`를 평가한다.
- 다음의 예시를 보자.

```python
1/0
```

```python
print(False and (1/0))  # 앞에꺼가 False니까 and에선 어차피 False -> 앞에꺼 반환 ********시험!!!!********
print(True or (1/0))    # 앞에꺼가  True니까  or에선 어차피  True -> 앞에꺼 반환
```

- 첫 번째 예시: 첫 번째 값 `False`만으로 결과가 결정되기 때문에, `(1/0)`은 실행되지 않는다.
- 두 번째 예시: 첫 번째 값 `True`만으로 결과가 결정되기 때문에, `(1/0)`은 실행되지 않는다.

```python
if True or (1/0):   # ** (1/0)때문에 에러가 나야 하는데 or 때문에 실행되버림! 조심!! **
    print("실행 됨")
else:
    print("실행 안됨")
```

- **** if가 True나 False를 판단할 필요가 없음. 그냥 True 자체가 반환되었기 때문에 ****

#### 3. `and`와 `or`가 `if`에 적용되는 경우
- `and`, `or`들의 반환값이 `if`의 조건식으로 사용되면, `if`는 그 결과값을 다시 `True` 또는 `False`로 판단한다.
- 이때, `and/or`은 Short-circuit evaluation을 수행하므로, 결과가 이미 결정되면 뒤의 피연산자는 평가하지 않는다.

```python
if False and 3:
    print("실행 됨")
else:
    print("실행 안됨")
```

```python
if False or 3:  # 3 반환 -> True로 인식
    print("실행 됨")
else:
    print("실행 안됨")
```

- `and`: 첫 번째 인자가 `False`이므로 뒤에 `3`을 평가하지 않고, `False`를 반환한다. 그렇기에, `실행 안됨` 결과를 반환한다.
- `or`: 첫 번째 인자가 `False`이므로 뒤에 `3`을 평가한다. 그러나, `3`은 `True`로 인식되므로, `if`는 최종 `True`로 인식한다. 따라서, `실행 됨`을 반환한다.

#### 4. `and/or`과 `&/|`의 차이
- `and`와 `&`, `or`과 `|`는 일부 경우에 같은 결과가 나타나지만, Python에서는 서로 다른 연산자이다.
- `bool`자료형 `True/False`에서는 동일한 결과를 얻을 수 있지만, 정수에서는 서로 다른 방식의 연산을 수행하며, 결결과가 달라질 수 있다. 일반적인 `list`, `tuple`에서는 `&/|`을 사용할 수 없다.
- `&/|`은 Short-Circuit evaluation을 하지 않는다.
- 정수에서 `&/|`을 사용하면, 숫자를 2진수로 표현한 뒤, 각 자리별로 연산한다.
- `&`는 각 자리에서 두 값이 모두 `1`인 경우에만 `1`이 된다.

```python
print(1 & 2) # 01 & 10 = 00 -> 0 (0*2^0 + 0*2^1)
print(3 & 1) # 11 & 01 = 01 -> 1 (1*2^0 + 0*2^0 = 1)
```

- `1`은 이진법으로 `01`, `2`는 이진법으로 `10`이다. 그러므로, `1 & 2`는 `01 & 10`의 결과인 `00`을 얻는다. 따라서, 정수 `0`이 반환된다.
- `3`은 이진법으로 `11`이다. 그러므로, `3 & 1`은 `11 & 01`의 결과인 `01`을 얻으므로, 정수인 `1`이 반환된다.

```python
if 1 & 2:
    print("된다")
else:
    print("안된다")
```

```python
if 1 and 2:
    print("된다")
else:
    print("안된다")
```

- `|`은 각 자리에서 두 값 중 하나 이상이 `1`이면 `1`이 된다.

```python
print(1 | 2) # 01 | 10 = 11 -> 3 (1*2^0 + 1*2^1 = 3)
```

- `1`은 `01`, `2`는 `10`이므로 `1 | 2`는 `11`이 되며, 이는 정수 `3`에 해당되므로 `3`을 반환한다.

```python
if False | 3:  # 0|3 = 00|11 = 11 -> 3 -> True
    print("실행 됨")
else:
    print("실행 안됨")
```

- `|`의 경우: `False | 3`이지만, `False`는 `0`처럼 동작하고, `3`은 2진수로 `11`이므로 `00 | 11`을 계산한 결과 `11`, 정수 `3`을 반환한다. 이후, `if`는 `3`을 `True`로 인식하므로 `실행 됨`을 최종 출력한다.
- `&`의 경우: `False & 3`이지만, `0 & 11`로 연산이 되고, `00`으로 연산되어 `0`을 반환한다. 이후, `if`에서 `0`을 `False`로 인식하므로 `실행 안됨`으로 최종 출력된다.
