## 5. Dictionary(딕셔너리)

- Dictionary는 `key`와 `value`의 쌍을 저장하는 자료형이다.
- Dictionary는 `{}`를 이용하여 생성하며, 각 원소는 `key: value` 형태로 구성한다.
- 각 원소는 `,`로 구분한다.

```
{key1: value1, key2: value2, ...}
```
- 다음 예시는 `key`로 1, `value`로 "hi"를 사용한 예이다.

````python
x = {1:'hi'}
x
````

- `value`에는 Number, String, List, Tuple 등 다양한 자료형의 객체를 저장할 수 있다.

````python
x = {'x': [1, 2, 3]}  # 여기서 x는 서로 다른 x임 / x: dictionary , 'x': key , [1, 2, 3]: value
x
````

### 1) Dictionary 원소 추가 및 삭제

- Dictionary에 새로운 `key`와 `value`의 쌍을 추가할 수 있다.
- `x[key] = value` 형태로 새로운 원소를 추가한다.

````python
x = {1: "a"}
print(x)

x[2] = "b"   # * 2라고 하는 숫자가 key  # 3번째 위치 나타내는 것 아님 *
print(x)
````

- 하나의 Dictionary에서 `key`와 `value`에 서로 다른 자료형을 사용할 수 있다.

````python
x['name'] = 'nada'  # key: 'name' ,  value: 'nada'
print(x)

x['tmp_list'] = [1, 2, 3]
print(x)
````

- `del` 문(statement)을 이용하여 Dictionary의 특정 원소를 삭제할 수 있다.
- `del x[key]`를 실행하면 해당 `key`와 연결된 `value`가 함께 삭제된다.

````python
del x[1]
print(x)
````

### 3) Dictionary 생성 시 주의사항
- Dictionary를 생성할 때 다음 사항을 주의해야 한다.
    1. 하나의 Dictionary에서 각 `key`는 고유해야 한다.
        - 동일한 `key`를 여러 번 사용하면 오류가 발생하지 않고, 마지막에 지정한 `value`가 저장된다.
    2. List는 Dictionary의 `key`로 사용할 수 없다.
        - List처럼 변경 가능한 객체는 `key`로 사용할 수 없다.
        - Number, String, Tuple 등은 `key`로 사용할 수 있다.

- 다음 예시를 살펴보자.

````python
x = {1:'a', 1 : 'c', 1:'b'}
x[1]  # 마지막 value 가져옴
````

- 동일한 `key`인 1이 여러 번 사용되었으므로 마지막에 지정한 `"b"`가 저장된 것을 확인할 수 있다.

````python
x = {[1, 2]: 'hi'}  # key를 list로 지정해서 에러남
````

- List `[1, 2]`는 `key`로 사용할 수 없어 `TypeError`가 발생하지만, Tuple `(1, 2)`는 `key`로 사용할 수 있다.

````python
x = {(1, 2): 'hi'}  # tuple은 가능
x[(1, 2)]
````

### 5) Dictionary 관련 method
- Dictionary에서도 다양한 method를 사용할 수 있다.
- method마다 사용할 수 있는 input과 반환되는 output이 다르다.
- 어떤 method는 새로운 값을 반환하고, 어떤 method는 기존 객체 자체를 변경한다.
- 따라서 method를 사용할 때는 input, output, 원본 객체의 변경 여부를 함께 확인해야 한다.

#### 1. keys()
- Dictionary에 저장된 모든 `key`를 `dict_keys` 형태의 객체로 반환한다.
- 기존 Dictionary는 변경되지 않는다.

````python
x = {'name': 'pey', 'phone': '010-9999-1234', 'birth': '1118'}
x.keys()
````

````python
type([1,2,3]) == type(x.keys())  # type([1,2,3])= list / type(x.keys()) -> 리스트가 아님!
````

#### 2. values()
- Dictionary에 저장된 모든 `value`를 `dict_values` 형태의 객체로 반환한다.
- 기존 Dictionary는 변경되지 않는다.

````python
x.values()
````

#### 3. items()
- Dictionary의 `(key, value)` 쌍을 `dict_items` 형태의 객체로 반환한다.
- 각 원소는 `(key, value)` 형태의 Tuple로 구성된다.
- 기존 Dictionary는 변경되지 않는다.

````python
x.items()  # list 아니고, dictionary item 임! # ('name', 'pey') / ('phone', '010-9999-1234') / ('birth', '1118') 이건 튜플
````

- `keys()`, `values()`, `items()`의 반환값은 `list()`를 이용하여 List로 변환할 수 있다.

````python
print(list(x.keys()))    # list로 변환
print(list(x.values()))  # 튜플들은 그대로고 안이 list임
print(list(x.items()))
````

#### 4. clear()
- Dictionary에 저장된 모든 원소를 삭제한다.
- 기존 Dictionary 객체 자체가 삭제되는 것은 아니며, 실행 후 빈 Dictionary `{}`가 된다.
- 기존 Dictionary를 변경하며 반환값은 `None`이다.

````python
x.clear()
print(x)   # {} 이건 반환값이 아님  # x가 dictionary 라는 정보는 살아있음
````

````python
del x   # 그냥 아예 없는 거. dictionary 남기고 뭐고 없음
````

#### 5. get(key, default)
- `get(key)`는 input으로 `key`를 받아 해당 `key`의 `value`를 반환한다.
- 해당 `key`가 존재하지 않으면 `None`을 반환한다.
- 두 번째 input인 `default`를 지정하면 `key`가 존재하지 않을 때 `None` 대신 `default`를 반환한다.
- 기존 Dictionary는 변경되지 않는다.

````python
x = {'name': 'pey', 'phone': '010-9999-1234', 'birth': '1118'}
print(x.get('name'))
print(x.get('phone'))
````

````python
x['name']
````

- `key`가 존재하는 경우 `x[key]`와 `x.get(key)`는 같은 `value`를 반환한다.
- `key`가 존재하지 않으면 `get()` method는 `None`을 반환하지만, `x[key]` 형태로 접근하면 `KeyError`가 발생한다.
- `None`은 값이 없음을 나타내는 객체이다.

````python
z = x.get('nokey')  # 에러가 안남  # default가 None
print(z)
````

````python
z = x['nokey']     # 에러남
````

- `key`가 존재하지 않을 때 `None`이 아닌 다른 값을 반환하고 싶다면 `get(key, default)` 형태로 두 번째 input을 지정한다.

````python
print(x.get('nokey', '없음'))
````

## Variable(변수)

### 1) Variable과 객체
- Variable은 객체를 가리키기 위해 사용하는 이름이다.
- `=`은 오른쪽의 객체를 왼쪽의 Variable에 대입하는 연산자이다.

```
a = 1
b = "python"
c = [1, 2, 3]
```

- 위 예시에서 `a`, `b`, `c`는 Variable이다.
- 예를 들어 `a = 1`을 실행하면 Variable `a`가 정수 객체 `1`을 가리키게 된다.
- Variable과 객체는 서로 같은 개념이 아니다.

````python
x = [1, 2, 3]
id(x)
````

- `x = [1, 2, 3]`을 실행하면 `[1, 2, 3]`이라는 List 객체가 생성되고, Variable `x`가 해당 객체를 가리키게 된다.
- `id()`는 객체를 구별하기 위한 고유한 정수 값을 반환하는 내장 함수이다.
- 두 Variable이 같은 객체를 가리키면 `id()`의 결과도 동일하다.
- `id()`의 결과는 실행 환경이나 실행 시점에 따라 달라질 수 있다.

### 2) List의 대입과 복사
- List를 다룰 때는 같은 객체를 가리키는 대입과 새로운 객체를 만드는 복사를 구분해야 한다.

#### 1. 같은 객체를 가리키는 대입

````python
y = x
print(id(y))
print(id(x))
````

- `y = x`는 새로운 List를 생성하는 것이 아니다.
- Variable `x`와 `y`가 동일한 List 객체를 가리키게 된다.
- 따라서 `id(x)`와 `id(y)`의 결과가 동일하다.
- 이 경우, 한 Variable을 통해 List를 변경하면 다른 Variable에서도 변경된 결과가 나타난다.

````python
x[0] = 4

print(x)
print(y)
````

````python
x = [1, 2, 3]
y = [1, 2, 3]

print(id(x))
print(id(y))
````

- `x`와 `y`가 동일한 List 객체를 가리키므로 `x`를 통해 첫 번째 원소를 변경하면 `y`에서도 변경된 결과가 나타난다.
- 기존 List와 값은 같지만 서로 다른 List 객체를 만들기 위해 Slicing이나 `copy()` method를 사용할 수 있다.

#### 2. Slicing을 이용한 복사

````python
z = x[:]

print([x, id(x)])
print([z, id(z)])
````

- `x[:]`는 `x`의 원소를 이용하여 새로운 List를 생성한다.
- 따라서 `x`와 `z`의 값은 같지만 서로 다른 객체이므로 `id()`의 결과가 다르다.

````python
x[0] = 4

print(x)
print(z)
````

- `x`와 `z`는 서로 다른 List 객체이므로 `x`의 첫 번째 원소를 변경해도 `z`에는 영향을 미치지 않는다.

#### 3. `copy()` method를 이용한 복사
- List의 `copy()` method를 이용하면 동일한 원소를 가진 새로운 List를 생성할 수 있다.

````python
x = [1, 2, 3]
y = x.copy()

print([x, id(x)])
print([y, id(y)])
````

- `y = x`와 달리 `y = x.copy()`는 새로운 List 객체를 생성한다.
- 따라서 `x`와 `y`의 값은 같지만 `id()`의 결과는 다르다.

### 3) 여러 Variable에 값 대입하기
- 여러 개의 값을 여러 Variable에 한 번에 대입할 수 있다.

````python
a, b = ('python', 'life')

print(a)
print(b)
````

- 위 문장은 다음과 같이 쓸 수도 있다.

```
(a, b) = 'python', 'life'
```

- Tuple은 괄호를 생략하여 표현할 수 있으므로 두 형태는 같은 의미이다.
- List 형태를 이용하여 여러 Variable에 값을 대입할 수도 있다.

````python
[a, b] = ["py", "li"]
print(a)
print(b)
````

- `=`을 연속해서 사용하면 여러 Variable에 동일한 객체를 대입할 수 있다.
- `x = y = "python"`을 실행하면 `x`와 `y`가 동일한 String 객체를 가리키므로 `id(x)`와 `id(y)`의 결과가 같다.

````python
x = y = "python"

print([x, y])
print([id(x), id(y)])
````
