## 1. $\pi$값 추정

- $\pi$는 3.141592....의 값을 가지는 무한소수이다.
- 반지름이 1인 원의 넓이는 $\pi$이다.
- 이때, $x, y \in [0, 1]$에서 $x^2 + y^2 \leq 1$을 만족하는 영역은 반지름이 1인 원의 $1/4$에 해당하므로, 넓이는 $\pi/4$이다.
- 따라서, $0, 1 \times [0, 1]$에서 $x$, $y$를 랜덤하게 생성하고, $x^2 + y^2 \leq 1$을 만족하는 비율을 계산하면 $\pi/4$를 근사할 수 있다.
- 이 비율에 4를 곱하면 $\pi$ 값을 근사할 수 있다.
    
1. $x, y$를 각각 uniform distribution with (0, 1)에서 생성한다.
2. $i = 1, 2, 3, ....$에 대해 다음을 계산한다.
    $$I_i = \left\{\begin{array}{cc} 1 & x^2 + y^2 \le 1 \\ 0 & \text{o.w} \end{array} \right.$$
3. $I_i$는 $i$번째 반복에서 생성된 점이 $x^2 + y^2 \leq 1$을 만족하는지를 나타내는 값이다. 그러므로, $$\frac{1}{n} \sum_{i = 1}^n I_i$$ 은 식을 만족하는 점의 비율을 의미한다.
4. (3.)에서 계산값이 $\pi/4$값과 충분히 가까워질 때까지 반복한다.

- 충분히 가까워질 때까지라는 표현은 두 값의 차이의 절대값이 일정한 기준값보다 작아지는 경우를 의미한다.
- 이 예시에서는 `while` statement의 종료 조건을 설정하기 위해 실제 $\pi$ 값을 이용한다.
- uniform distribution에서 랜덤하게 값을 생성하기 위해 `random` 모듈의 `uniform` 함수를 사용하고, $\pi$ 값은 `math` 모듈의 `pi`를 사용한다.
- 이는 다음과 같이 구현할 수 있다.

```python
## 추정값 자체가 필요할 때!!!
import random
import math

# 3. 점 갯수 변수
count = 0
# 4. 비교값
true_value = math.pi/4
# 5. 총 횟수
total_iter = 0


# 8. 반복
while True:
    # 1. 점 생성
    x = random.uniform(0,1)
    y = random.uniform(0,1)

    # 2. 원 안에 들어가냐 안들어가냐
    count = count + ((x**2 + y**2) <=1) # 처음엔 1 또는 0밖에 안나옴


    # 6. 총횟수 업데이트
    total_iter = total_iter + 1

    # 7. 가까워졌냐 (근사되냐)    # abs(count/ total_iter - true_value): 이 값이 0에 가까울수록 실제값에 가까움
    # 9. 조건 입력(if_break)
    if abs(count/ total_iter - true_value) <= 1e-3:
        break

print([total_iter, count/total_iter - true_value])
```

```python
# 1,0 데이터 필요할 때!!!
import random
import math

i = 1
ind = []

while(1):
    x = random.uniform(0, 1)
    y = random.uniform(0, 1)

    ind = ind + [(x**2 + y**2 <= 1)]

    if(abs(sum(ind)/len(ind) - math.pi/4) <= 10e-5):
        break

    i = i + 1


[i, sum(ind)/len(ind), sum(ind)/len(ind)*4, math.pi]
```

- 각 반복이 진행됨에 따라 $\pi/4$의 값과 유사한 값이 나오게 되고, 그 값에 4를 곱한값이 $\pi$값으로 추정된것을 확인 할 수 있다.

## 2. 구분구적법  # 생략함
- 구분구적법은 실수공간을 정의역으로 가지는 임의의 함수 $f(x)$에 대해 함수의 넓이를 계산하는 방법이다.
- 이 넓이를 계산하기 위해, `x`축을 임의의 `n`개로 구분하고, 그 구간의 높이와 곱한 사각형의 넓이를 더하는 방법이다. 이때, `n`을 무한히 보내면, 실제 적분값으로 수렴한다.
- $x \in [a, b]$에서 함수 $f(x)$에 대한 구분구적법을 표현하면 다음과 같다.
    $$\lim_{n \rightarrow \infty} \sum_{i = 1}^n \frac{b - a}{n} \left( \frac{f(x_i) + f(x_{i - 1})}{2}\right).$$
이다. 이때, $x_i$들은
$$a = x_0 < x_1 < \dots < x_{n - 1} < x_n = b$$
을 만족한다.
- 예시로 구간 $[0, 1]$에서 $\exp(x)$함수를 적분하면
    $$\int_0^1 \exp(x) dx = \exp(1) - 1 \approx 1.718282$$
임을 알 수 있다.
- 0이 아닌 정수 $n$에 대해 구분구적법 수식으로 표현하면 다음과 같다.
    $$\lim_{n \rightarrow \infty} \sum_{i = 1}^n \frac{1}{n}\left( \frac{\exp(x_i) + \exp(x_{i - 1})}{2} \right)$$
이때, $x_0 = 0 < x_1 < x_2 < \dots < x_{n - 1} < x_n = 1$이다.
- $t$번째 반복에서 이를 추정하는 순서는 다음과 같다. <p>
    
1. $[0, 1]$구간을 $t$개의 구간으로 분할하여 $x_1, x_2, \dots, x_{n - 1}$을 얻는다.
2. $x_i$들에 대해 $\exp(x_i)$값들을 계산한다.
3. 구분구적법 수식을 계산하여 $\exp(1) - 1$의 값과 비슷할때까지 반복한다. <p>

- 이는 다음과 같이 구현 할 수 있다. 이때, `numpy` 모듈의 `exp` 함수를 이용하여 `exponential function`의 값을 계산하고, `linspace` 함수를 이용하여 구간을 동일한 간격으로 나눈다.

```python
import numpy as np

i = 1
true_value = np.exp(1) - 1

while(1) :
    seq_vec = np.linspace(0, 1, num = i)
    exp_value = np.exp(seq_vec)

    tmp_value = (exp_value[:(i - 1)] + exp_value[1:])/2
    mean_value = sum(tmp_value)/i

    if abs(mean_value - true_value) <= 1e-4 :
        break

    i = i + 1

[i, mean_value, true_value]
```

## 3. Monte Carlo method를 이용한 기댓값 추정    **중요!!!**
- 랜덤하게 생성한 값을 반복적으로 이용하여 원하는 값을 근사하는 방법을 Monte Carlo method라 한다.
- 이 방법은 확률변수의 기댓값을 계산하는데 많이 사용된다.
- 확률변수 $X$와 함수 $g(X)$를 생각하자.
- $X_1, X_2, \dots, X_n$이 서로 독립이고 동일한 distribution에서 생성되었다고 하자.
- 만약, $E[|g(X)|]$가 존재하는 경우, 대수의 법칙(Law of Large Number)에 의해 다음이 성립한다.
$$ \lim_{n \to \infty} \frac{1}{n} \sum_{i = 1}^n g(X_i) \to E[g(X)].$$
- 따라서, 기댓값을 직접 계산하기 어려운 경우, 이 방법을 이용하여 기댓값을 추정할 수 있다.
- 확률변수 $X$가 Standard Normal distribution $N(0, 1)$을 따른다면, $E[X^2] = 1$을 만족한다.
- $E[X^2]$을 추정하기 위해, 다음의 순서를 진행한다.
    1. 반복횟수 $n$을 정한다.
    2. $X_i$를 Standard Normal distribution에서 생성한다.
    3. $X_i^2$을 계산하여 누적한다.
    4. (2.), (3.) 과정을 $n$번 반복한다.
    5. $X_1^2, \dots, X_n^2$의 평균을 계산하여 $E[X^2]$을 추정한다.
- Standard Normal distribution에서 랜덤하게 값을 생성하기 위해, `random`모듈의 `gauss`함수를 사용한다.
- 아래 코드는 $n$이 `500`까지 반복하며 10번에 한번씩 추정값을 보여주는 코드이다.

- 출력값은 `[반복 횟수, 추정값, 실제값과의 차이]`를 의미한다.
- 반복횟수가 증가하면서, 추정값이 이론값 1에 가까워 지는 경향이 발생한다.
- 다만, 랜덤하게 생성된 값을 이용하므로 반복 횟수가 증가할 때마다 오차가 항상 감소하는 것은 아니다.
- 대수의 법칙은 표본평균이 반복 횟수의 증가에 따라 기댓값에 수렴한다는 것을 의미하며, 매 단계에서 단조롭게 가까워진다는 의미는 아니다.
