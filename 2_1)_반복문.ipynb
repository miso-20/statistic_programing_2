# 반복문
- 조건문 뿐만아니라, 프로그래밍에 있어 가장 중요한 구문 중 하나이다.
- 단순 반복을 진행하는 구문이며, `for`statement을 이용한다.
- 기본적인 형태는 다음과 같다.
```
for Variable in 반복 가능한 객체:
    statement 1
```
- `for` statement의 `in` 뒤에는 반복 가능한 객체가 온다.
- `List`, `Tuple`, `String`, `range` 등이 반복 가능한 객체에 해당한다.
- 반복 가능한 객체의 원소가 순서대로 `Variable`에 대입되며, 각 원소마다 들여쓰기된 문장이 실행된다.
- 다음의 예시는 `1`부터 `5`까지 반복하여 출력하는 코드이다.

````python
for i in [1, 2, 3, 4, 5]:  # ***** 암기!!****
    print(i)
````

1. `i`에 1부터 5까지 하나씩 할당 받는다.
2. 할당 받은 `i`를 출력한다.

- 다음 예시는 `List x`의 각 원소에 자기 자신을 더하여, 각 원소를 2배로 변경하는 코드이다.

````python
x = [1, 2, 3, 4, 5]

for i in [0, 1, 2, 3, 4]:
    x[i] = x[i] + x[i]   # x[0] = x[0]+x[0] = 1+1 =2
                         # x[1] = x[1]+x[1] = 2+2 =4 ....
                         # x[4] = x[4]+x[4] = 5+5 =10

x
````

- 반복되는 원소가 `Tuple`인 경우, `Tuple`의 각 원소를 여러 `Variable`에 나누어 대입할 수 있다.
- 다음 예시에서는 `(1, 2), (3, 4), (5, 6)`의 각 원소를 `first, second`에 각각 대입한다.

````python
for (first, second) in [(1, 2), (3, 4), (5, 6)]:
    print(first + second)
````

````python
x = [(1, 2), (3, 4), (5, 6)]
for (first, second) in x:
    print(first + second)
````

````python
x = [(1, 2), (3, 4), (5, 6)]
for [first, second] in x:  # list도 가능
    print(first + second)
````

1. `(first, second)`에 튜플을 할당 받는다.
2. 할당받은 튜플의 원소에 대해 합을 계산하여 출력한다.

## 1. 조건문 적용
- 5개의 숫자에 대해, 60이상의 갯수를 세는 코드를 조건문을 이용해서 다음과 같이 구현 할 수 있다.

````python
x = [90, 25, 67, 45, 80]

(x[0] >= 60) + (x[1] >= 60) + (x[2] >= 60) + (x[3] >= 60) + (x[4] >= 60)
````

````python
x = [90, 25, 67, 45, 80]

num_con = 0
for tmp_x in x:
    if tmp_x >= 60:
        num_con = num_con + 1  # x=90, 67, 80 일때, 1씩 증가 = 3
    else :
        num_con = num_con      # x=25, 45 일때, 변화 X

print(num_con)
````

1. `num_con`은 60 이상인 값의 개수를 저장하는 `Variable`이다.
2. `x`의 원소 하나를 `tmp_x`에 할당한다.
3. 할당된 `tmp_x`가 `60`보다 크거나 같은지 확인한다.
4. 크거나 같다면 `num_con`에 1을 더하고, 아닐 경우 그대로 할당한다.

- 같은 방법으로, 값이 60 이상이면 `"over"`, 60 미만이면 `"under"`를 저장하는 `List`를 생성해 보자.

````python
y = [0, 0, 0, 0, 0]   # 위의 x = [90, 25, 67, 45, 80]

for i in [0, 1, 2, 3, 4]:
    if x[i] >= 60:
        y[i] = "over"
    else:
        y[i] = "under"

print(y)
````

1. `over`, `under`을 저장할 리스트 `y`를 생성한다.
2. `i`에 `0`부터 `4`까지 순차적으로 할당된다.
3. 리스트 `x`에서 `i`번째 원소를 가져와 `60`보다 크거나 같은지 조건을 확인한다.
4. 조건을 만족하면 `List y`의 i번째 원소에 `"over"`를 저장하고, 만족하지 않으면 `"under"`를 저장한다.

- 위 예시를 리스트의 `+`연산을 이용하면 다음과 같이 생성 할 수 있다.

````python
y = []     # 위의 x = [90, 25, 67, 45, 80]

for i in [0, 1, 2, 3, 4]:
    if x[i] >= 60:
        y = y + ["over"]
    else:
        y = y + ["under"]

print(y)
````

````python
y = []     # 위의 x = [90, 25, 67, 45, 80]

for i in [0, 1, 2, 3, 4]:
    if x[i] >= 60:
        y.append("over")   # y=ppend("over")  이렇게 쓰면 에러남! **
    else:
        y.append("under")  # y=ppend("under") 이렇게 쓰면 에러남! **

print(y)
````

## 2. 같이 사용하는 statement 및 function
- `for` statement와 같이 사용하면 좋은 statement 및 `function`, `method`들을 소개한다.

#### 1. continue statement
- `for` statement안에서 `continue`를 만나면 현재 반복에서 남아 있는 코드를 실행하지 않고, 다음 반복으로 넘어간다.
- 다음의 예시는 `60`이하가 몇개인지 확인하는 코드이다.

````python
x = [90, 25, 67, 45, 80]

num = 0
for tmp_x in x:
    if tmp_x <= 60:
        num = num + 1
    else:
        continue

num
````

- 위 코드에서 `num = num + 1`코드를 추가해서 돌려보자.

````python
x = [90, 25, 67, 45, 80]

num = 0
for tmp_x in x:
    if tmp_x <= 60:
        num = num + 1
    else:
        num = num + 1

num
````

````python
x = [90, 25, 67, 45, 80]

num = 0
for tmp_x in x:
    if tmp_x <= 60:
        num = num + 1 # x=25,45에서만 + 1
    else:
        continue  # x=90이어서 넘어왔는데 continue 때문에 다음 문장 실행 안하고 위로 다시 올라감-> x=25부터 반복
        num = num + 1 # continue 때문에 수행 안함

num
````

- 그래도 똑같이 `2`가 출력되는 것을 확인할 수 있다.
- `tmp_x`가 `60`보다 큰 경우 `continue`를 만나면 현재 반복에서 남아 있는 코드를 실행하지 않고 다음 반복으로 넘어간다.
- 따라서, `continue` 아래에 있는 `num = num + 1`은 실행되지 않는다.

#### 2. range function    *****암기!!!!*****
- `range` 함수는 일정한 규칙을 가지는 정수의 범위를 나타내는 `range` 객체를 생성하는 파이썬 내장 함수이다.
- `range` 함수를 이용하여 생성한 변수는 데이터 타입이 `range`이다.
- 사용법은 다음과 같다.
    - `range(x)` : `0`부터 `x - 1`까지의 정수 범위를 나타낸다.
    - `range(x, y)` : `x`부터 `y - 1`까지의 정수 범위를 나타낸다.
    - `range(x, y, z)` : `x`부터 시작하여 `z`만큼 증가하면서 `y`에 도달하기 전까지의 정수 범위를 나타낸다.

````python
print(range(5))
print(type(range(5)))
````

- `range` 객체를 `List`로 변환하려면 `list()` 함수를 사용할 수 있다.

````python
print(list(range(5)))  # ***list로!***
print([range(5)])
````

- `list(range(5))`는 `range` 객체의 원소를 이용하여 새로운 `List`를 생성한다.
- `[range(5)]`는 `range(5)` 객체 자체를 하나의 원소로 가지는 `List`를 생성하므로 두 결과는 서로 다르다.
- `for` statement와 `range` 함수를 이용하면, 다음과 같은 구구단을 생성 할 수 있다.

````python
for i in range(2, 10):
    for j in range(1, 10):
        print(i*j, end = " ")
    print("")
````

#### 3. append method
- `append` method는 `input`을 `List`의 마지막에 하나의 원소로 추가한다.
- 기존 `List` 자체를 변경하며 반환값은 `None`이다.
- 반복문에서 결과를 하나씩 `List`에 저장할 때 사용할 수 있다.

````python
x = [1, 2, 3, 4]
y = []

for tmp_x in x:
    y.append(tmp_x*3)

y
````

- 이를 리스트 연산을 이용하여 똑같이 생성할 수 있다.

````python
x = [1, 2, 3, 4]
y = []

for tmp_x in x:
    y = y + [tmp_x*3]

y
````

- `+` 연산자와 `append` method는 `List`에 원소를 추가하는 방식에 차이가 있다.
- `y = y + [tmp_x*3]`은 기존 `List y`와 새로운 `List [tmp_x*3]`을 연결하여 새로운 `List`를 생성한 후 다시 `y`에 저장한다.
- `y.append(tmp_x*3)`은 기존 `List y`에 `tmp_x*3`을 하나의 원소로 추가하며, 기존 `List` 자체를 변경한다.
- 따라서, 위 예시에서는 두 방법의 결과는 같지만 `List`를 처리하는 방식에는 차이가 있다.
