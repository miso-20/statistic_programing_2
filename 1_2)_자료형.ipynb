## 3. List(리스트)
- List(리스트)는 여러 개의 값을 하나의 객체에 저장하는 자료형이다
- List는 `[]`를 이용하여 생성하고, 각 원소는 `,`로 구분한다.
- List의 원소에는 숫자, String, List, Tuple 등 다양한 자료형을 저장할 수 있다.

```python
# +)
a = ["a,b","c"]
print(a)
```

```python
a = [1, 2, 3, 4]
print(a)

a = ["a", "b", "c"]
print(a)

a = [1, 2, "a", "b"]
print(a)

a = [1, 2, ["a", "b"]]
print(a)
```

### 1) List의 Indexing(인덱싱)
- List도 String과 동일한 방식으로 Indexing하며, index는 0부터 시작한다.

```python
a = [1, 2, 3]
print(a)

print(a[0])
```

- `-1`의 의미는 마지막 원소를 의미한다.

```python
print(a[-1])
```

- List의 원소로 또 다른 List를 저장할 수 있다. 중첩된 List의 원소에 접근하려면 Indexing을 연속해서 사용한다.

```python
a = [1, 2, ["a", "b"]]
a[-1][0] # 마지막 원소 ["a", "b"]의 1번째 원소
a[-1][1]
```

```python
type(a[2]) # 3번째 ["a", "b"] : list
```

```python
type(a[1])
```

### 2) List의 Slicing(슬라이싱)
- List의 Slicing(슬라이싱)은 String과 동일한 방식으로 사용하며, 일정 범위의 원소를 가져올 수 있다.

```python
a = [1, 2, 3, 4, 5]

print(a[0:2])
print(a[2:4])
print(a[:2])
print(a[2:])
```

### 3) List 연산
- List에서는 `+`, `*` 연산을 사용할 수 있다.
- `+` : 두 List를 연결하여 새로운 List를 반환한다.
- `*` : List를 지정한 횟수만큼 반복한 새로운 List를 반환한다.
- ** 두 연산 모두 기존 List 자체를 변경하지 않는다. **

```python
x = [1, 2, 3]
y = [4, 5, 6]

x + y  # *순서 중요 # 사칙연산 아님!! *
```

```python
## +)
y + x
```

```python
[1, 2, 3]*4
```

```python
x*4
```

### 4) List 수정 및 삭제
- List는 생성한 후에도 원소를 변경할 수 있다.
- Indexing을 이용하여 특정 위치의 원소에 새로운 값을 저장할 수 있다.
- List의 원소에는 다양한 자료형의 값을 저장할 수 있다.

```python
x = [1, 2, 3]
print(x)

x[1] = 4   # 2번째 값에 4 대입
x
```

```python
x = [1, 2, 3]
print(x)

x[0] = (4, 5)
x
```

- `del` 문(statement)을 이용하면 객체의 특정 원소 또는 객체를 가리키는 이름을 삭제할 수 있다.

```python
del x[0]
x
```

- `del x[0]`은 List x의 첫 번째 원소를 삭제한다.
- `del` statement는 List의 원소뿐만 아니라 변수 자체를 삭제할 수도 있다.

```python
del x
x      # 없는데 가져오라 했으니까 에러남
```

### 6) List 관련 method
- method마다 사용할 수 있는 input과 반환되는 output이 다르다.
- 어떤 method는 새로운 값을 반환하고, 어떤 method는 기존 객체 자체를 변경한다.
- 따라서 method를 사용할 때는 input, output, 원본 객체의 변경 여부를 함께 확인해야 한다.
  
#### 1. append(x)
- input으로 받은 x를 List의 마지막에 하나의 원소로 추가한다.
- ** 이때 기존 List 자체가 변경되며, 반환값은 `None`이다 **

```python
x = [1, 2, 3]
result = x.append(4)
print(x)
print(result)
```

```python
# +)
x.append(5)
print(x)
```

- append()의 input에는 하나의 객체가 들어가며, 해당 객체 전체가 하나의 원소로 추가된다

```python
x.append((5, 6))
x
```

#### 2. extend(x)
- `x`에 포함된 여러 원소를 하나씩 List의 마지막에 추가한다.
- 기존 List 자체를 변경하며, 반환값은 None이다.
- List뿐 아니라 Tuple 등 여러 원소를 순서대로 제공할 수 있는 객체를 input으로 사용할 수 있다.

```python
x = [1, 2, 3]
x.extend([4, 5]) # 리스트로 저장됨
x
```

```python
x.append([4,5]) # 리스트 안의 리스트*로 저장됨
print(x)
```

```python
x = [1, 2, 3]
```

```python
x + [4, 5, 6]
```

```python
x
```

```python
x.extend([4,5,6])
x
```

- `extend()`와 `+` 연산은 List를 연결한다는 점은 유사하지만, `+` 연산은 새로운 List를 반환하고 `extend()`는 기존 List 자체를 변경한다.
####  3. sort()
- List의 원소를 정렬한다.
- 기존 List 자체를 변경하며, 반환값은 None이다.

```python
x = [1, 4, 3, 2]
x.sort()
x
```

- String도 정렬할 수 있으며, String의 비교 규칙에 따라 순서가 결정된다.

```python
x = ["a", "d", "b", "c"]
x.sort()
x
```

#### 4. reverse()
- 현재 List 원소의 순서를 반대로 뒤집는다.
- 기존 List를 변경하며, 반환값은 None이다.

```python
x = [1, 2, 3, 4]
x.reverse()  # x = x.reverse() 이렇게 사용하면 안됨
x
```

```python
x = [1, 2, 3, 4]
result = x.reverse()
print(result)  # 반환값은 result에만 저장됨
x
```

#### 5. index(x)
- `index(x)`는 input으로 값 `x`를 받아, List에서 `x`가 처음 나타나는 위치의 index를 output으로 반환한다

```python
x = [1, 2, 3]
print(x.index(1)) # '1'이 위치하는 곳이 어디냐
print(x.index(3))
```

- 1이라는 값은 List에서는 1번째 위치지만, 파이썬에서는 Indexing이 0부터 시작하기 때문에 0을 반환한다. 또한, 3은 3번째이지만, 2을 반환한다.
- input으로 받은 값이 List에 존재하지 않으면 output을 반환하지 못하고 ValueError가 발생한다

```python
x.index(0)
```

#### 6. insert(x, y)
- `insert(x, y)`는 두 개의 input을 사용한다.
    - `x` : 값을 추가할 index
    - `y` : 추가할 값
- 기존 List를 변경하며 반환값은 None이다.

```python
x = [1, 2, 3]
x.insert(0, 4)  # 0번째(실제 1번째)에 4를 삽입
x
```

- 두 번째 input인 y에는 다양한 자료형의 객체가 들어갈 수 있다

```python
x = [1, 2, 3]
x.insert(3, (4, 5))
x
```

#### 7. remove(x)
- input으로 값 `x`를 받아, 처음 나타나는 `x`를 삭제한다.
- 기존 List 자체를 변경하며 반환값은 None이다.

```python
x = [1, 2, 3, 1, 2, 3]
x.remove(3)  # 첫 번째 '3'만 삭제
```

#### 8. pop(x)
- index `x`의 원소를 삭제하고, 삭제한 원소를 output으로 반환한다.
- input을 생략하면 마지막 원소를 삭제하고 반환한다.

```python
x = [1, 2, 3]
result = x.pop()  # ** 순서 중요- 삭제 하고, 반환하고**
print(x)
print(result)
```

```python
x = [1, 2, 3]
result = x.pop(0) # 2,3 이 저장됨
print(x)
print(result)     # 1을 return
```

#### 9. count(x)
- input으로 값 `x`를 받아 List에서 `x`가 나타나는 개수를 정수 형태의 output으로 반환한다

```python
x = [1, 2, 3, 1, 2]
x.count(1)
```

## 4. Tuple(튜플)
- Tuple(튜플)은 List와 유사하게 여러 개의 값을 저장하는 자료형이다.
- List와 달리 Tuple은 생성한 후 원소를 추가, 삭제, 수정할 수 없다.
- List는 `[]`, Tuple은 일반적으로 `()`를 이용하여 생성한다.
- 원소가 하나인 Tuple은 값 뒤에 `,`를 사용해야 한다.

```python
x = (1, 2, "a", "b")
print(x)
print(type(x))
```

- Tuple은 `()`를 생략하여 생성할 수도 있다.

```python
x = 1, 2, "a", "b"
print(x)
print(type(x))
```

- Tuple은 생성된 후 원소를 삭제하거나 수정할 수 없다. 이를 시도하면 TypeError가 발생한다.

```python
del x[0]
```

```python
x[0] = 5
```

- `del x[0]`은 Tuple의 원소를 삭제하려는 것이므로 불가능하지만, `del x`는 Tuple을 변경하는 것이 아니라 변수 `x` 자체를 삭제하는 것이므로 실행할 수 있다.

```python
del x # tuple 없기 때문에 에러 발생 # 위의 list del x 의 에러와 다름
x
```

- `(1)`은 Tuple이 아니라 정수 `1`을 괄호로 묶은 표현이다. 원소가 하나인 Tuple을 생성하려면 `(1, )`처럼 `,`를 사용해야 한다

```python
x = [1]
print(type(x))

y = (1)
print(type(y))

z = (1, )
print(type(z))
```

### 1) Indexing 및 Slicing
- Tuple의 특징은 원소를 변화 시킬수 없다는 것만 제외하면 나머지는 List와 같은 방법으로 다룬다.

```python
x = (1, 2, "a", "b")
print(x[0])
print(x[3])
print(x[0:2])
print(x[:3])
```

### 2) Tuple 연산
- List와 마찬가지로 `+`, `*`연산을 이용한다.
    - `+` : 두 Tuple을 연결한 새로운 Tuple을 반환한다.
    - `*` : Tuple을 지정한 횟수만큼 반복한 새로운 Tuple을 반환한다.
- 기존 Tuple 자체는 변경되지 않는다.

```python
x = (1, 2, "a", "b")
y = (3, 4)

z = x + y
print(z)
```

```python
z = y*2
z
```

- `+` 연산으로 두 객체를 연결하려면 같은 자료형끼리 연산해야 한다. 예를 들어 Tuple은 Tuple과, List는 List와 `+` 연산을 할 수 있다.
- +) *(곱하기)는 자료형 상관없음

### 3) Tuple 길이 구하기
- Tuple에서도 len()을 사용할 수 있으며, input으로 Tuple을 받아 원소의 개수를 정수 형태의 output으로 반환한다.

```python
x = (1, 2, 3)
len(x)
```
