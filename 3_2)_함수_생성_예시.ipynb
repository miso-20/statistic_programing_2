## 1. 절대값 함수
- input : 실수 또는 정수, output : 절댓값
- 문자형을 입력하면 안내 문장을 출력하고 `None`을 반환한다.
- type(x): 입력값이 문자형인지 확인할 수 있다.

```python
def my_abs(x) :
    if type(x) == str:
        print("문자형입니다. 숫자를 넣어주세요.")
        # return_value =  None 코드에선 동치관계
        return None

    else:
        return_value = x*(x >= 0) + x*(x < 0)*(-1)

    return(return_value)
```

```python
my_abs(-5)
```

```python
my_abs(1)
```

```python
my_abs("10")
```

## 2. 중복값 제거 함수
- input : 숫자로 구성된 리스트, output : 중복된 값을 제거한 리스트
- 리스트 원소가 처음 등장한 순서를 유지한다.

```python
x = [1, 1, 2, 2, 3, 4, 1, 2, 3, 4, 5]
x
```

- 코드를 실행하고, 저장한다
  1. 엎어쓴다. -> 적용할 수 없음. 기존 정보가 존재해야 하기 때문에
  2. 새로 만든다. -> 적용: uni_x list
- uni_x: unique한 값을 저장하는 공간
- uni_x에 값이 없으면: 새로운 unique한 값
- uni_x에 있으면: 새로운 unique가 아닌 값
- 위의 2개 단계를 x의 원소에 대해 반복
- GPT나 정보를 찾았을 때, in operator를 이용한다.
- list안에 값이 있으면 True, 없으면 False
    - uni_x = [1, 2, 3] : 가정
    - uni_x in 1 => True  => uni_x = uni_x
    - uni_x in 4 => False => uni_x = uni_x + [4]

```python
x = [1, 1, 2, 2, 3, 4, 1, 2, 3, 4, 5]

def my_unique(x):
    uni_x = []

    for ele in x:
        if ele not in uni_x: # ele가 있으면 그냥 넘어감
            uni_x.append(ele) # 1: 빈공간이니까 1을 여기 넣음

    return uni_x

my_unique(x)
```

- uni_x = [] 할당
- 첫 번째 for: ele에 1이 할당 -> ele not in uni_x하면 True: uni_x는 빈공간이고, 1은 그 안에 없으니까
-> 조건이 만족되니까, uni_x에 1을 저장: append -> uni_x = [1]
- 두 번째 for: ele에 1이 할당 -> ele not in uni_x하면 False: uni_x = [1]이니까 -> 조건 만족 안 됨, 넘어감
- 세 번째 for: ele에 2가 할당 -> ele not in uni_x하면  True: uni_x = [1]이니까 -> 조건 만족 -> uni_x에 2를 저장: append -> uni_x = [1, 2]
- 순서를 언급한 이유: 앞에서부터 비교하면서 지우기 때문에
