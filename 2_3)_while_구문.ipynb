# while statement
- `while` statement는 주어진 조건식이 `True`인 동안 문장을 반복해서 실행하는 반복문이다.
- `for` statement는 반복 가능한 객체의 원소를 순서대로 사용하여 반복하고, `while` statement는 조건식이 `True`인 동안 반복한다.
- 조건식이 `False`가 되면 `while` statement가 종료된다.
- 기본적인 형태는 다음과 같다.
```
while 조건식:
    statement 1
```
- 조건식이 `True`이면 들여쓰기된 문장을 실행한 후 다시 조건식을 확인한다.
- 조건식이 `False`이면 `while` statement를 종료한다.
- `for`문에서 1부터 5까지 반복하며 출력 하는 예시를 진행했었다. 이 예시를 `while`로 만들면 다음과 같다.

````python
x = 1

while x <= 5:
    print(x)
    x = x + 1
````

````python
x = 1

while x <= 5:
    print(x)
    x = x + 1

print("====================")
print(x)
````

1. `x = 1`을 할당한다.
2. 조건식 `x <= 5`가 `True`이므로 `x`를 출력한다.
3. `x = x + 1`을 실행하여 `x`에 `1`을 더한다.
4. 다시 조건식 `x <= 5`를 확인하고, `True`이면 같은 과정을 반복한다.
5. `x = 6`이 되면 조건식 `x <= 5`가 `False`가 되므로 `while` statement가 종료된다.  **** **`x = 6`이 되면******

## 1. 반복문 제어 statement
- `while` statement는 조건식이 `True`인 동안 반복을 계속한다.
- 따라서 조건식이 계속 `True`이면 반복문이 종료되지 않으며, 이를 무한루프(infinite loop)라고 한다.
- 다음 예시는 종료 조건이 없으므로 계속해서 반복된다.
```
x = 1

while True :
    print(x)
    x = x + 1
```
#### 1. break statement
- `break` statement는 현재 실행 중인 반복문을 즉시 종료할 때 사용한다.
- 특히 `while True`와 같이 반복을 계속하도록 설정한 경우, 특정 조건에서 반복문을 종료하기 위해 사용할 수 있다.
- 위의 무한루프 코드에서, `x`가 20까지 출력되는 예시는 다음과 같다.

````python
x = 1
while True:
    print(x)
    x = x + 1

    if x == 21:
        break
````

- `x`가 `1`씩 증가하며 출력된다.
- `x = 20`을 출력한 후 `x`에 `1`을 더하면 `x = 21`이 된다.
- 이때, `x == 21`이 `True`가 되어 `break`가 실행되고, `while` statement가 종료된다.
- 다음 예시를 생각해보자. `x = 1`에서 시작하여 `x`를 반복할 때마다 2배씩 증가시킨다.
- 이때, `x`가 처음으로 `1e+15` 이상이 되는 시점이 몇 번째 반복인지 확인해보자.

````python
x = 1
count = 1

while True:
    x = x*2
    print([count, x])
    count = count + 1

    if x >= 1e+15:
        break
````

````python
x = 1
count = 0

while True:
    x = x*2
    count = count + 1
    print([count, x])

    if x >= 1e+15:
        break
````

- 반복 횟수를 미리 정하기 어려운 경우에는 `while` statement를 사용할 수 있다.
- 위 예시는 `x`가 `1e+15` 이상이 될 때까지 반복하므로, 종료 조건을 기준으로 반복하는 `while` statement를 이용하여 구현할 수 있다.

#### 2. continue statement
- `continue` statement를 만나면 현재 반복에서 남아 있는 문장을 실행하지 않고 다음 반복으로 넘어간다.
- 다음 예시는 10 이하의 숫자 중 짝수만 `List`에 저장하는 과정이다.

````python
x = 0
y = []

while True:
    x = x + 1

    if x % 2 != 0:
        continue

    y = y + [x]  # [] 꼭 써야함

    if x >= 10 :
        break

print(y)
````

- `x`에 `1`을 더한 후 `x > 10`인지 확인한다. `x > 10`이면 `break`를 실행하여 `while` statement를 종료한다.
- `x % 2 != 0`이 `True`이면 `x`는 홀수이므로 `continue`를 실행하고 다음 반복으로 넘어간다.
- 따라서, 홀수는 `List y`에 저장되지 않고, 짝수인 경우에만 `y = y + [x]`가 실행된다.
- 최종적으로 `y`에는 `[2, 4, 6, 8, 10]`이 저장된다.
