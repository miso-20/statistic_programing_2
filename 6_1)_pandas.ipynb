# Pandas
- Pandas는 데이터를 처리하고 분석하기 위해 사용하는 Python 패키지로, 주로 DataFrame이라는 2차원 데이터 구조를 다룬다.
- Excel, CSV, 텍스트 파일 등 다양한 형태의 데이터를 읽고, 정리하고, 분석에 필요한 형태로 변환하는 과정을 Data Handling이라고 한다.
- Pandas는 이러한 Data Handling을 수행하기 위한 다양한 기능을 제공한다.
- 참고 : https://pandas.pydata.org/docs/getting_started/intro_tutorials/index.html

```python
import pandas as pd
```

## 1. Dataframe(데이터 프레임)
- DataFrame의 행(row)과 열(column)은 각각 다음과 같은 특징을 가진다.
    - 행(row): 하나의 관측 대상(observation)에 대한 정보를 나타낸다.
    - 열(column): 하나의 변수(variable) 또는 특성(characteristic)에 대한 정보를 나타내며, 일반적으로 동일한 자료형의 값들로 구성된다.
    - 하나의 column에 숫자들이 있지만, 문자로 인식한다. ex) "1,000" -> format 개념
- 각 행에는 이를 구분하는 `인덱스(index)`가 있으며, 각 열에는 `열 이름(column name)`이 있다.
- 인덱스와 열 이름에는 숫자나 문자 등을 사용할 수 있다.

### 1-1) DataFrame 정의
- DataFrame은 pd.DataFrame()을 이용하여 생성할 수 있다.
- 아래 코드는 5명의 이름, 성별, 나이, 키를 저장한 DataFrame을 생성하는 예제이다.

```python
df = pd.DataFrame(
    { 'name' : ['이철수', '김영희', '홍길동', 'John Smith', 'Mary Doe'],
      'sex'  : ['M', 'F', 'M', 'M', 'F'],
      'age'  : [ 23, 25, 21, 33, 45],
      'height' : [153.5, 175.3, 163.4, 180.0, 165.7]
    } )

print(df)
print(type(df))
```

- DataFrame을 생성하는 방법 중 하나는 `dictionary`를 이용하는 것이다.
    - Dictionary의 key는 열 이름(column name), value는 해당 열에 저장되는 값들을 나타낸다.
    - 예를 들어 'name': ['이철수', '김영희', '홍길동', 'John Smith', 'Mary Doe']는 name이라는 열을 생성하고, 5명의 이름을 저장한다.
- Dictionary에서 각 항목은 쉼표(,)로 구분한다.
- `type`함수를 이용하여 확인해보면, pandas의 DataFrame 객체임을 알 수 있다.

### 1-2) Indexing 및 Slicing
- DataFrame에서도 NumPy 배열과 마찬가지로 Indexing과 Slicing을 이용하여 원하는 데이터를 선택할 수 있다.
- DataFrame에서는 다음과 같은 방법으로 데이터를 선택할 수 있다.
    - `열 이름(column name)`을 이용하여 특정 열을 선택한다.        ## slicing
    - `loc` method를 이용하여 행과 열의 이름을 기준으로 선택한다.  ## slicing
    - `iloc` method를 이용하여 행과 열의 위치를 기준으로 선택한다. ## indexing
- Boolean filtering 이용하여 조건을 만족하는 데이터를 선택한다. 이때, `[]` 또는 `loc` method를 이용한다.

```python
print(df["name"], '\n')
print(df.name)
```

- df.종가(원)가 아니라 df["종가(원)"] 으로 써야 인식됨

```python
sub_df = df[["name", "sex"]]
print(sub_df)
```

```python
df[["name", "sex"]]
```

```python
df[df['height'] <= 165]
```

```python
print(df['sex'] == 'M')
print((df['height'] >= 160.0))
print((df['sex'] == 'M') & (df['height'] >= 160.0))
```

```python
df[(df['sex'] == 'M') & (df['height'] >= 160.0)]
```

```python
df.loc[(df['height'] >= 170.0) , 'name']
```

```python
df.loc[ (df['sex'] == 'M') , ['name', 'height']]
```

```python
df.iloc[0,0]
```

```python
df.iloc[3,1:3] # 이름 빼고 / 성별,나이,이름 가져옴
```

```python
df.iloc[0,:]
```

### 1-3) Series
- `DataFrame`을 구성하는 하나의 row, column은 `Series`란 반환되는 경우가 있다.
